```python
import numpy as np

a = np.arange(11, 30, 2)
print(a)
print("Dimensions:", a.ndim)
print("Total elements:", a.size)
```

```python
a = np.linspace(5, 35, 7)
print(a)
```

```python
a = np.full((4, 3), 9)
a[2, 1] = 0
print(a)
```

```python
a = np.array([6, 12, 18, 24])
print(a % 5)
```

```python
a = np.array([14, 28, 42, 56, 70, 84])
print(a[1:6:2])
```

```python
a = np.array([-16, 25, -36, 49])
b = np.abs(a)
print(b)
print(np.sqrt(b))
```

```python
a = np.arange(7, 13).reshape(3, 2)
print(a)
print("Original shape:", a.shape)
print(a.T)
print("Transposed shape:", a.T.shape)
```

```python
temperatures = np.array([18, 22, 27, 31, 24, 19])

print("Range:", np.max(temperatures) - np.min(temperatures))
print("Median:", np.median(temperatures))
```

```python
values = np.array([4, 9, 4, 2, 9, 7, 2, 4])

a = np.unique(values)
print(a)
print("Count:", len(a))
```

```python
speed = np.array([35, 62, 48, 81, 55])

result = np.where(speed >= 60, "Fast", "Normal")
print(result)
```

```python
import pandas as pd
s = pd.Series([24, 31, 19, 27],
              index=['Monday', 'Tuesday', 'Wednesday', 'Thursday'])
print(s)
```

```python
s = pd.Series([45000, 18000, 22000],
              index=['Laptop', 'Tablet', 'Phone'])
print(s['Tablet'])
```

```python
rainfall = pd.Series([12, 0, 18, 5, 23])
print(rainfall[rainfall > 10])
```

```python
data = {'Gold': 7200, 'Silver': 95, 'Copper': 850}
s = pd.Series(data)

print(s.index)
print(s.values)
```

```python
ratings = pd.Series([3.5, 4.2, 2.8, 4.7])
ratings.name = 'Customer Ratings'
ratings.index.name = 'Review ID'

print(ratings)
```

```python
s = pd.Series([11, 22, 33], index=['X1', 'X2', 'X3'])

print('X2 exists:', 'X2' in s.index)
print('X4 is absent:', 'X4' not in s.index)
```

```python
stock = pd.Series([16, 28, 12],
                  index=['Keyboard', 'Mouse', 'Monitor'])

print(stock.to_dict())
```

```python
data = {'Hindi': 78, 'English': 84, 'Computer': 91}
s = pd.Series(data, index=['Computer', 'Hindi', 'Art', 'English'])

print(s)
print("Missing entry:", s.isnull()['Art'])
```

```python
expenses = pd.Series([1200, None, 850, None, 600])

print(expenses.isnull())
print(expenses.notnull())
```

```python
a = pd.Series([8, 15, 22], index=['P', 'Q', 'R'])
b = pd.Series([4, 6, 10], index=['R', 'P', 'Q'])

print(a + b)
```
